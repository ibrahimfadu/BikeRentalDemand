```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression, ElasticNet
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import KFold

sns.set_style('whitegrid')
%matplotlib inline

print("Libraries loaded ✅")
```

## 4. RMSLE Metric

$$\text{RMSLE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(\log(p_i+1) - \log(a_i+1))^2}$$

```python
def rmsle(y_true, y_pred):
    """Root Mean Squared Logarithmic Error."""
    y_pred = np.maximum(y_pred, 0)
    return np.sqrt(np.mean((np.log1p(y_pred) - np.log1p(y_true)) ** 2))

print("RMSLE function defined ✅")
```

```python
def engineer_features(df, keep_targets=False):
    df = df.copy()
    df['datetime'] = pd.to_datetime(df['datetime'])
    df['month']     = df['datetime'].dt.month
    df['dayofweek'] = df['datetime'].dt.dayofweek
    df['hour']      = df['datetime'].dt.hour

    is_weekend   = df['dayofweek'] >= 5
    weekday_peak = (~is_weekend) & (df['hour'].isin([7,8,9,17,18,19]))
    weekend_peak = (is_weekend)  & (df['hour'].between(10, 18))
    df['peak_hour'] = (weekday_peak | weekend_peak).astype(int)

    drop_cols = ['datetime', 'holiday', 'temp']
    if not keep_targets:
        drop_cols += [c for c in ['casual','registered','count'] if c in df.columns]

    return df.drop(columns=[c for c in drop_cols if c in df.columns])

train_full = engineer_features(train, keep_targets=True)
train_fe   = train_full.drop(columns=['casual','registered','count'])

X = train_fe
y = train_full['count']
feature_cols = list(X.columns)

print(f"Features ({len(feature_cols)}):", feature_cols)
print(f"X shape: {X.shape}, y shape: {y.shape}")
```

```python
def cross_validate(model, X, y, n_splits=10, seed=42):
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=seed)
    scores = []
    for tr, val in kf.split(X):
        model.fit(X.iloc[tr], y.iloc[tr])
        preds = model.predict(X.iloc[val])
        scores.append(rmsle(y.iloc[val], preds))
    return float(np.mean(scores)), float(np.std(scores))

print("CV helper ready ✅")
```

```python
## 7. Train All 8 Models
```

```python
results = {}

# ---- 1. Linear Regression ----
print("1. Linear Regression ...")
s, std = cross_validate(LinearRegression(), X, y)
results['Linear'] = s
print(f"   RMSLE: {s:.4f} (+/- {std:.4f})")

# ---- 2. GLMNet (ElasticNet) ----
print("2. GLMNet (ElasticNet) ...")
glmnet = Pipeline([
    ('scaler', StandardScaler()),
    ('enet', ElasticNet(alpha=0.1, l1_ratio=0.5, max_iter=5000))
])
s, std = cross_validate(glmnet, X, y)
results['GLMNet'] = s
print(f"   RMSLE: {s:.4f} (+/- {std:.4f})")

# ---- 3. PCR ----
print("3. PCR ...")
pcr = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA(n_components=0.95)),
    ('lr', LinearRegression())
])
s, std = cross_validate(pcr, X, y)
results['PCR'] = s
print(f"   RMSLE: {s:.4f} (+/- {std:.4f})")

# ---- 4. SVR ----
print("4. SVR (RBF) ...")
svr = Pipeline([
    ('scaler', StandardScaler()),
    ('svr', SVR(kernel='rbf', C=10, epsilon=0.5))
])
s, std = cross_validate(svr, X, y)
results['SVR'] = s
print(f"   RMSLE: {s:.4f} (+/- {std:.4f})")
```

```python
# ---- 5. Random Forest ----
print("5. Random Forest ...")
rf = RandomForestRegressor(n_estimators=200, max_depth=15,
                           random_state=42, n_jobs=-1)
s, std = cross_validate(rf, X, y)
results['Random Forest'] = s
print(f"   RMSLE: {s:.4f} (+/- {std:.4f})")

# ---- 6. GBM ----
print("6. GBM ...")
gbm = GradientBoostingRegressor(n_estimators=200, learning_rate=0.05,
                                max_depth=5, random_state=42)
s, std = cross_validate(gbm, X, y)
results['GBM'] = s
print(f"   RMSLE: {s:.4f} (+/- {std:.4f})")

# ---- 7. CTree (DecisionTree substitute) ----
print("7. CTree (DT substitute) ...")
ctree = DecisionTreeRegressor(max_depth=12, min_samples_leaf=5,
                              random_state=42)
s, std = cross_validate(ctree, X, y)
results['CTree (DT sub.)'] = s
print(f"   RMSLE: {s:.4f} (+/- {std:.4f})")
```

```python
# ---- 8. Stacking Ensemble ----
print("8. Stacking (RF + CTree -> Linear meta) ...")
kf = KFold(n_splits=10, shuffle=True, random_state=42)
stack_scores = []

for tr, val in kf.split(X):
    inner = int(0.8 * len(tr))
    tr_main, tr_meta = tr[:inner], tr[inner:]

    m_rf = RandomForestRegressor(n_estimators=200, max_depth=15,
                                 random_state=42, n_jobs=-1)
    m_ct = DecisionTreeRegressor(max_depth=12, min_samples_leaf=5,
                                 random_state=42)
    m_rf.fit(X.iloc[tr_main], y.iloc[tr_main])
    m_ct.fit(X.iloc[tr_main], y.iloc[tr_main])

    meta_X = np.column_stack([m_rf.predict(X.iloc[tr_meta]),
                              m_ct.predict(X.iloc[tr_meta])])
    meta_model = LinearRegression().fit(meta_X, y.iloc[tr_meta])

    val_X = np.column_stack([m_rf.predict(X.iloc[val]),
                             m_ct.predict(X.iloc[val])])
    stack_scores.append(rmsle(y.iloc[val], meta_model.predict(val_X)))

results['Stacking'] = float(np.mean(stack_scores))
print(f"   RMSLE: {np.mean(stack_scores):.4f} "
      f"(+/- {np.std(stack_scores):.4f})")
```

```python
# ---- BONUS: Split regression ----
print("BONUS: Split (registered + casual) ...")
y_reg, y_cas = train_full['registered'], train_full['casual']
split_scores = []

for tr, val in kf.split(X):
    rf_reg = RandomForestRegressor(n_estimators=200, max_depth=15,
                                   random_state=42, n_jobs=-1)
    rf_cas = RandomForestRegressor(n_estimators=200, max_depth=15,
                                   random_state=42, n_jobs=-1)
    rf_reg.fit(X.iloc[tr], y_reg.iloc[tr])
    rf_cas.fit(X.iloc[tr], y_cas.iloc[tr])
    pred = rf_reg.predict(X.iloc[val]) + rf_cas.predict(X.iloc[val])
    split_scores.append(rmsle(y.iloc[val], pred))

results['Split (reg+cas)'] = float(np.mean(split_scores))
print(f"   RMSLE: {np.mean(split_scores):.4f}")
```
